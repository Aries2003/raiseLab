Translation Cleaning


```python
import pandas as pd

path = "/home/valmeid1/raiseLab/output/Translation/deepseek-r1_processed.parquet"
df = pd.read_parquet(path)

think_mask = df['model_translation'].str.contains(r'<think>|</think>', regex=True, na=False)
df.loc[think_mask, 'model_translation'] = (
    df.loc[think_mask, 'model_translation']
    .str.split(r'</think>', regex=True)
    .str[-1]
    .str.strip()
)

df.to_parquet(path, index=False)
```
