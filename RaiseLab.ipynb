```python
import pandas as pd

parquet_file_path = '/home/valmeid1/raiseLab/output/Translation/deepseek-r1.parquet'
df_parquet = pd.read_parquet(parquet_file_path, engine='pyarrow')

file_name = parquet_file_path.split('/')[-1]
print(f"File Name: {file_name}")

print("\nDataFrame Head:")
display(df_parquet.head())
```

```python
if 'id' in df_parquet.columns:
    df_parquet = df_parquet.drop(columns=['id'])

df_parquet['id'] = range(1, len(df_parquet) + 1)

print("\nDataFrame with new 'id' column:")
display(df_parquet.head())
```

```python
output_parquet_file_path = '/home/valmeid1/raiseLab/output/Translation/deepseek-r1_processed.parquet'
df_parquet.to_parquet(output_parquet_file_path, index=False)

print(f"Processed DataFrame saved to: {output_parquet_file_path}")
```
